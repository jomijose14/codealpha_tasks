#### Step 9 : 
__________________________________________________________________________________________________________________-
__________________________________________________________________________________________________________________
* Merging both datasets is essential for better data analysis.

* The frequency column needs to be standardized, as it contains variations like **' Monthly', 'Monthly', and ' M'.**

* The **Area** column is not present in **unemp_df2**, while the **general_region** column is missing in **unemp_df1**, requiring adjustments during merging.

* Since both **Rural and Urban** data exist for the same state and date, they will be combined by taking the mean value instead of dropping any entries to prevent significant data loss.

* The Area (Rural & Urban) column in unemp_df1 will be dropped, as it is not available in **unemp_df2**. Instead, the **general_region column (South, North, West, Northeast, etc.)** will be used.
#### ____________________________________________________________________________________________________________________



```python
# Standardizing the frequency column
unemp_df1['frequency'] = unemp_df1['frequency'].str.strip().replace({'M': 'Monthly'})
unemp_df2['frequency'] = unemp_df2['frequency'].str.strip().replace({'M': 'Monthly'})
```

```python
 #Handling Missing Columns Before Merging
# Add 'area' column to unemp_df2 with NaN values
unemp_df2['area'] = None  # or np.nan
unemp_df1['general_region'] = None  # or np.nan
```

```python
# Create a mapping of state to general_region from unemp_df2
state_region_mapping = unemp_df2.set_index('state')['general_region'].to_dict()

# Add 'general_region' column to unemp_df1 using the mapping
unemp_df1['general_region'] = unemp_df1['state'].map(state_region_mapping)

# Fill missing values if any
unemp_df1.loc[:, 'general_region'] = unemp_df1['general_region'].fillna('Unknown')
```

```python
# Create mappings for state → longitude and state → latitude from unemp_df2
state_to_longitude = unemp_df2.set_index('state')['longitude'].to_dict()
state_to_latitude = unemp_df2.set_index('state')['latitude'].to_dict()

# Map longitude and latitude to unemp_df1
unemp_df1['longitude'] = unemp_df1['state'].map(state_to_longitude)
unemp_df1['latitude'] = unemp_df1['state'].map(state_to_latitude)

# Fill missing values with NaN or 'Unknown'
unemp_df1['longitude'] = unemp_df1['longitude'].fillna(np.nan)
unemp_df1['latitude'] = unemp_df1['latitude'].fillna(np.nan)
```

```python
# Step 11: Remove the "Area" column
unemp_df1_copy= unemp_df1.copy()
df_cleaned = unemp_df1_copy.drop(columns=['area'])
```

```python
# Step 12: Here, we are combining Rural & Urban area together based on state & date by calculating mean value of unemployment & labour participation rate and sum of estimated employment.
# Group by Region, Date, and Frequency, then aggregate appropriately
df_cleaned = df_cleaned.groupby(['state', 'date', 'frequency','general_region','longitude','latitude'], as_index=False).agg({
    'unemployment_rate': 'mean',  # Averaging unemployment rate
    'estimated_employed': 'sum',  # Summing employed population
    'labour_participation_rate': 'mean'  # Averaging labour participation rate
})
```

```python
# Step 12: Here, we are identifying the unique values from the first dataframe by comparing with the second dataframe.
#Then combine the unique entries from the first file and second file. 
df1_unique = df_cleaned[~df_cleaned.set_index(['state', 'date']).index.isin(
    unemp_df2.set_index(['state', 'date']).index
)]

# Save new data to a CSV file
#df1_unique.to_csv("new_data.csv", index=False)

# Step 3: Combine unemp_df2 (all data) and df1_unique (only new data)
df_combined = pd.concat([unemp_df2, df1_unique], ignore_index=True)

# Save the final dataset
df_combined.to_csv("final_combined_data.csv", index=False)

print(f" New data saved as 'new_data.csv' ({len(df1_unique)} records).")
print(f" Final combined dataset saved as 'final_combined_data.csv' ({len(df_combined)} records).")
```

COVID-19 led to severe disturbances, with unemployment rates soaring from 8% to 24% during the lockdown period from March to May 2020. Additionally, labour force participation dropped from 44% to 35%. The four phases of the nationwide lockdown, which lasted from March 25 to May 31, 2020, directly correlated with the most severe impacts on the labour market.

**Regional Differences:**

- South: Highest unemployment spike (43.58%) but relatively stronger pre/post-lockdown.

- North: Persistently high unemployment (14.97%) and lowest participation rate.

- Northeast: Highest labour participation (52.93%) despite elevated unemployment.

- West: Maintained lower unemployment rates throughout the crisis.

**State Performance:**

- *Puducherry and Tamil Nadu* : Most dramatic unemployment spikes (74.65% and 43.43%).

- *Tripura and Haryana* : Consistently high unemployment across all periods.

- *Meghalaya, Odisha, Assam* : Greater resilience with lower unemployment.

- *Tripura, Meghalaya, Telangana* : Highest sustained labour participation.

Many areas have not fully recovered after the lockdown. More people are working than before, but the labour participation rate is still lower than it was before the pandemic. Unemployment rates have gone down from their highest points, but they often remain higher than they were before the crisis. The strength of recovery varies a lot by region.

The pandemic changed how the labour market usually works. Before the lockdown, there were strong links between unemployment, employment, and participation. These links weakened during the lockdown and did not fully return to normal by the end of the observation period.

**Future Work Required:**

- Sector Analysis: Examine different economic sectors' contributions to unemployment patterns and identify which industries showed greater resilience.

- Demographic Impact: Investigate employment effects across demographic groups (age, gender, education, urban/rural) and identify disproportionately affected segments.

- Policy Assessment: Analyse government intervention effectiveness and compare outcomes in states with different economic support measures.

- Resilience Factors: Determine which economic, social, and governance factors contributed to greater resilience in certain regions.

- Extended Monitoring: Continue tracking unemployment trends beyond initial recovery to assess return to pre-pandemic trajectories.

- Skills Gap Evaluation: Assess pandemic-driven changes in skill requirements and develop recommendations for education and training programs.
